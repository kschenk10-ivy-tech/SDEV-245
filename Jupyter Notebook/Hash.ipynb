### 2. Hash a string
- **Strings:** encode to bytes with `.encode("utf-8")`.
<br>
`hashlib.sha256()` expects **bytes**, so first the text needs to be encoded as UTF-8.
<br>
`.hexdigest()` formats the digest as readable hexadecimal.

```python
import hashlib

def hash_string(text):
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

print(hash_string("hello"))
```

### 3. Hash a file

- **Files:** read raw bytes with `open(path, "rb")`.

Open the file in **binary mode** (`"rb"`) and feed its contents to the hasher.<br>
Reading in chunks means even large files can be processed without loading them entirely into memory.

```python
def hash_file(path):
    hasher = hashlib.sha256()
    with open(path, "rb") as file:
        for chunk in iter(lambda: file.read(65536), b""):
            hasher.update(chunk)
    return hasher.hexdigest()
```

## A Simple, Interactive App

Run this cell, enter `1` for text or `2` for a file, and follow the prompt. For files, supply a path accessible to the notebook’s Python process.

```python
def main():
    choice = input("Hash (1) text or (2) file? ").strip()

    if choice == "1":
        text = input("Enter text: ")
        print("SHA-256:", hash_string(text))
    elif choice == "2":
        path = input("Enter file path: ").strip()
        try:
            print("SHA-256:", hash_file(path))
        except OSError as error:
            print("Could not read file:", error)
    else:
        print("Please enter 1 or 2.")

main()
```

### Verification

The SHA-256 digest of `abc` is a well-known test value. The assertions below confirm that string hashing works and that file hashing returns the same digest for identical bytes.

```python
from pathlib import Path

expected = "ba7816bf8f01cfea414140de5dae2223b00361a396177a9cb410ff61f20015ad"
assert hash_string("abc") == expected

test_path = Path("example.txt")
test_path.write_bytes(b"abc")
try:
    assert hash_file(test_path) == expected
finally:
    test_path.unlink()

print("Both tests passed!")
```
