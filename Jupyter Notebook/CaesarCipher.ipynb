# Encoding and decoding using the Caesar Cipher

*This Python script shifts each letter forward or backward through the alphabet using ASCII number conversions and modular arithmetic, while leaving punctuation and spaces untouched.*


```python
def caesar_cipher(text, shift, mode='encode'):
    if mode == 'decode':
        shift = -shift
    
    result = []
    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            # Shift character and wrap around the 26-letter alphabet
            shifted = (ord(char) - base + shift) % 26 + base
            result.append(chr(shifted))
        else:
            # Leave spaces, numbers, and punctuation unchanged
            result.append(char)
            
    return ''.join(result)

# --- Example Usage ---
message = "Hello, World! 123"
shift_amount = 3

encrypted = caesar_cipher(message, shift_amount, mode='encode')
decrypted = caesar_cipher(encrypted, shift_amount, mode='decode')

print(f"Original:  {message}")
print(f"Encrypted: {encrypted}")
print(f"Decrypted: {decrypted}")
```
